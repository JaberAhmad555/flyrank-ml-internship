## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains 30,000 pages, and 54.21% of them are labeled as declining. However, only 17 pages satisfy the simple rule of being both stale (not updated for at least 180 days) and still visible (at least 500 impressions in the last 90 days). This suggests that a simple stale-and-visible rule may be too narrow to prioritize all potentially important pages. A more flexible scoring approach using several observable signals may therefore be useful for building a ranked content-review queue.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
declining = df["trend_direction"].str.lower().eq("down")
stale_visible = (
    (df["days_since_last_update"] >= 180) &
    (df["impressions_90d"] >= 500)
)
print("Total pages:", len(df))
print("Declining rate:", round(declining.mean() * 100, 2), "%")
print("Stale and visible pages:", stale_visible.sum())
```
